Realizamos la Simulación en Código

```python
from pickle import FLOAT
import matplotlib.pyplot as plt
import control as ct
import numpy as np

s = ct.TransferFunction.s
R = 1000 #Valor de la Resistencia
C = 0.00006 #Valor de la Capacitancia
Hcnum = [1]
Hcden = [R*C*1, 1]
Hc = ct.tf(Hcnum, Hcden)
Hc
```

```python
# Simulación de la respuesta al escalón unitario
t, y = ct.step_response(Hc, T=1)

# Graficar
plt.figure()
plt.plot(t, y, label='Planta H(s)')
plt.xlabel('Tiempo (s)')
plt.ylabel('Salida')
plt.title('Respuesta al escalón unitario de la planta')
plt.grid()
plt.legend()
plt.show()
```
